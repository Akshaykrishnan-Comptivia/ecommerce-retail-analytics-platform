## **1. Create Catalog**

```sql
%sql
CREATE CATALOG IF NOT EXISTS ecommerce_analytics_catalog
MANAGED LOCATION 's3://databricks-storage-7474650621272587/unity-catalog/7474650621272587';
```

```sql
%sql
SHOW CATALOGS;
```

## **2. Set Active Catalog**

```sql
%python
spark.sql("USE CATALOG ecommerce_analytics_catalog")
print("Catalog set to: ecommerce_analytics_catalog")
```

## **3. Create Schema**

3.1 Bronze Schema

```sql
%sql
CREATE SCHEMA IF NOT EXISTS ecommerce_analytics_catalog.bronze
COMMENT 'Raw landing zone - no transformations applied';
```

3.2 Silver Schema

```sql
%sql
CREATE SCHEMA IF NOT EXISTS ecommerce_analytics_catalog.silver
COMMENT 'Cleansed, typed, deduplicated data with DQ flags';
```

3.3 Gold Schema

```sql
%sql
CREATE SCHEMA IF NOT EXISTS ecommerce_analytics_catalog.gold
COMMENT 'Normalized star schema dimensions, facts, and data mart';
```

##  4. Create Raw Data Volume

```sql
%sql

CREATE VOLUME IF NOT EXISTS ecommerce_analytics_catalog.bronze.raw_data
COMMENT 'Raw landing-zone files for synthetic and public source data';
```

## 5. Create Checkpoints Volume

```sql
%sql
CREATE VOLUME IF NOT EXISTS ecommerce_analytics_catalog.bronze.checkpoints
COMMENT 'Checkpoint files for Bronze, Silver, and Gold processing';
```

```sql
%sql
SHOW SCHEMAS IN ecommerce_analytics_catalog;
```
