## Product Tesing

```sql
%python
from src.silver.transform_products import transform_products

bronze_df = spark.table("ecommerce_analytics_catalog.bronze.bronze_products_csv")

silver_df = transform_products(bronze_df)

display(silver_df)

silver_df.printSchema()

print(f"Bronze Count : {bronze_df.count()}")
print(f"Silver Count : {silver_df.count()}")
```

```sql
%python
display(silver_df.filter("is_valid = false"))
```

```sql
%python
display(silver_df.filter("is_valid = false").select("product_id","product_weight_g","dq_reason"))
```

```sql
%python
display(bronze_df.filter(col("product_weight_g") <= 0))
```

```sql
%python
print("Bronze:", bronze_df.count())
print("Silver:", silver_df.count())
print("Invalid:", silver_df.filter("is_valid = false").count())
```

```sql
%python
from pyspark.sql.functions import col

display(silver_df.groupBy("product_id").count().filter(col("count") > 1))
```

```sql
%python
print("Bronze Products")
display(bronze_df.limit(10))

print("Silver Products")
display(silver_df.limit(10))
```
